# The Concept: How Git Works

---


Git moves your changes through four areas:

**Working Directory** -> **Staging Area** -> **Local Repo** -> **Remote (GitHub)**

```mermaid
sequenceDiagram
    participant WD as Working Directory
    participant SA as Staging Area
    participant LR as Local Repo
    participant R as Remote (GitHub)
    WD->>SA: git add
    SA->>LR: git commit
    LR->>R: git push
    R->>LR: git fetch
    LR->>WD: git pull
```

Three things to remember:
1. Save often (`git commit`)
2. Push to remote (`git push`)
3. Branch for experiments (`git checkout -b experiment`)


# Visual: Commit History (DAG)

---


````python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(10, 5))
ax.axis('off')

# Main branch commits
main_commits = [0, 1, 2, 3, 4]
main_y = [2] * 5

# Feature branch commits (diverges from main commit 1)
feature_commits = [1.5, 2.5, 3.5]
feature_y = [4] * 3

# Draw main branch line
ax.plot(main_commits, main_y, 'o-', color='#2c3e50', linewidth=3, markersize=10, label='main')

# Draw feature branch line (starts at main commit 1)
ax.plot([1, 1.5, 2.5, 3.5], [2, 4, 4, 4], 'o-', color='#e74c3c', linewidth=3, markersize=10, label='feature/new-optimizer')

# Draw merge back to main
ax.plot([3.5, 4], [4, 2], 'o-', color='#27ae60', linewidth=3, markersize=10)

# Labels
ax.annotate('main', xy=(0, 2.4), ha='center', fontweight='bold', fontsize=11, color='#2c3e50')
ax.annotate('branch created', xy=(1, 3.2), ha='center', fontsize=9, color='#e74c3c')
ax.annotate('experiment commits', xy=(2.5, 4.4), ha='center', fontsize=9, color='#e74c3c')
ax.annotate('merged back to main', xy=(4, 1.6), ha='center', fontsize=9, color='#27ae60')

ax.set_xlim(-0.5, 4.5)
ax.set_ylim(0.5, 5)
ax.set_title('Git Commit History: Branch & Merge', fontsize=14, fontweight='bold', pad=20)
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
````

> **Key insight:** Branches let you experiment without touching `main`. When the experiment works, merge it back. When it fails, just delete the branch — `main` stays clean.

# Step 2: The Daily Workflow

---


Save your work and back it up to GitHub:

```bash
git status
git add file.py
git commit -m "Add perceptron implementation"
git push origin main
```

| Step | Command | What it does |
|------|---------|--------------|
| 1 | `git status` | See which files changed |
| 2 | `git add <file>` | Stage changes for the next commit |
| 3 | `git commit -m "message"` | Save a snapshot with a message |
| 4 | `git push origin main` | Upload to GitHub |


# Visual: Why .gitignore Matters for AI Files

---


````python
import matplotlib.pyplot as plt

file_types = ['Source Code (.py)', 'Config (.yaml, .json)', 'Model Checkpoint (.pt)', 'Model Checkpoint (.pth)', 'Model Weights (.safetensors)']
sizes_kb = [5, 2, 500000, 2000000, 5000000]  # in KB
sizes_mb = [s / 1024 for s in sizes_kb]

colors = ['#27ae60', '#27ae60', '#e74c3c', '#e74c3c', '#e74c3c']

fig, ax = plt.subplots(figsize=(9, 4))
bars = ax.bar(file_types, sizes_mb, color=colors, alpha=0.8)
ax.set_ylabel("File Size (MB)", fontsize=11)
ax.set_title("Typical File Sizes: Why Model Checkpoints Belong in .gitignore", fontsize=12, fontweight="bold")
ax.set_yscale('log')
ax.grid(True, alpha=0.3, axis='y')
plt.xticks(rotation=15, ha='right')

for bar, mb in zip(bars, sizes_mb):
    ax.text(bar.get_x() + bar.get_width()/2, mb * 1.1, f"{mb:,.0f} MB", ha="center", fontweight="bold", fontsize=9)

plt.tight_layout()
plt.show()
````

> **Model checkpoint files (.pt, .pth, .safetensors) are often gigabytes.** Git is not designed for large binary files.
Add them to `.gitignore` and regenerate or store them separately.


# Step 3: Branching for Experiments

---


Create a branch, experiment, then merge back to `main`:

```bash
# Create and switch to a new branch
git checkout -b experiment/new-optimizer

# ... make changes, test, commit ...
git add .
git commit -m "Try new optimizer"

# Switch back to main
git checkout main

# Merge the experiment into main
git merge experiment/new-optimizer
```

If the experiment fails, you can abandon it:
```bash
git checkout main
git branch -D experiment/new-optimizer
```

# Key Terms

---


| Term | What people say | What it actually means |
|------|----------------|----------------------|
| Commit | "Saving" | A snapshot of your entire project at a point in time |
| Branch | "A copy" | A pointer to a commit that moves forward as you work |
| Merge | "Combining code" | Taking changes from one branch and applying them to another |
| Remote | "The cloud" | A copy of your repo hosted somewhere else (GitHub, GitLab) |
| Staging Area | "Preparing" | The intermediate area where you select changes before committing |


# Quiz Review

---


Here are the quiz questions from the actual lesson. Try to answer them before looking at the explanations.

**Q1:** What does 'version control' primarily help you do?
> **Answer:** Track changes to files over time and collaborate with others. Version control systems like git track every change made to files, allowing you to revert to previous states and collaborate with others on the same codebase.

**Q2:** What is a 'repository' in the context of software development?
> **Answer:** A collection of files and their complete change history managed by version control. A repository (repo) is a directory tracked by git that contains all project files along with the full history of changes.

**Q3:** What is the correct sequence for saving and backing up your work in git?
> **Answer:** `git add`, `git commit`, `git push`. First you stage changes, then save a snapshot, then upload to the remote.

**Q4:** What does `git checkout -b experiment/new-optimizer` do?
> **Answer:** Creates a new branch named `experiment/new-optimizer` and switches to it. The `-b` flag creates and switches immediately.

**Q5:** Why should you add `.pt`, `.pth`, and `.safetensors` to your `.gitignore`?
> **Answer:** They are model checkpoint files that are too large for git and can be re-generated. Git is not designed for large binary files.
