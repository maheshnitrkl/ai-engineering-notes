# Shapes Must Match

---

Matrix multiplication has a strict rule: `(m × n) @ (n × p) = (m × p)`. The **inner dimensions must match**.

```
(128 × 784) @ (784 × 1) = (128 × 1)
  weights      input      output

Inner dimensions: 784 = 784  ✓ valid
```

Shape mismatch errors in PyTorch come from violating this rule.


# Element-Wise vs Matrix Multiplication

---

These are **different operations with different rules** — mixing them up is a common bug.

**Element-wise:** multiply matching positions. Both arrays same shape.
```
| 1  2 |   | 5  6 |   |  5  12 |
| 3  4 | × | 7  8 | = | 21  32 |
```

**Matrix multiply:** dot products of rows × columns. Inner dims match.
```
| 1  2 |   | 5  6 |   | 19  22 |
| 3  4 | @ | 7  8 | = | 43  50 |
```

# Key Terms

---

| Term | What it means |
|------|---------------|
| Matrix multiply | Dot products between every row of the first matrix and every column of the second. Order matters. |
| Transpose | Swap rows and columns — turns m×n into n×m. Critical in backpropagation. |
| Determinant | Measures how much the matrix scales area/volume. Zero = transformation crushes a dimension. |
| Broadcasting | Stretching a smaller array to match a larger one. |
| Identity matrix | The matrix equivalent of multiplying by 1 — used in residual connections (ResNets). |
