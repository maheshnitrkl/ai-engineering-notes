# Key Concepts

---



### Why shapes matterMatrix multiplication has a strict rule: `(m × n) @ (n × p) = (m × p)`.The inner dimensions must match.```(128 × 784) @ (784 × 1) = (128 × 1)  weights       input       outputInner dimensions: 784 = 784  -- valid```### Element-wise vs matrix multiplicationElement-wise: multiply matching positions. Both matrices must be the same shape.Matrix multiplication: dot products of rows and columns. Inner dimensions must match.### BroadcastingStretching a smaller array to match a larger one by repeating along missing dimensions.

````python
import numpy as npA = np.array([[1, 2], [3, 4]])B = np.array([[5, 6], [7, 8]])print("A + B =")print(A + B)print("A @ B (matrix multiply) =")print(A @ B)print("A^T =")print(A.T)print("det(A) =", np.linalg.det(A))# Dense layer: relu(W @ x + b)inputs = np.random.randn(3, 1)weights = np.random.randn(2, 3)bias = np.array([[0.1], [0.1]])output = np.maximum(0, weights @ inputs + bias)print(f"Layer output shape: {output.shape}")
````
