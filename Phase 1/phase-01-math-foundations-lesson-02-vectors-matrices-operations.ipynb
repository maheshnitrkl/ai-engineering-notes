# The Problem

---

You want to build a neural network. You read the code and see this:

```
output = activation(weights @ input + bias)
```

That `@` is matrix multiplication. The `weights` are a matrix. The `input` is a vector. If you do not know what those operations do, this line is magic. If you do know, it is the entire forward pass of a layer in three operations.


# Shapes Matter

---

Matrix multiplication has a strict rule: `(m × n) @ (n × p) = (m × p)`. The inner dimensions must match.

```
(128 × 784) @ (784 × 1) = (128 × 1)
  weights      input      output

Inner dimensions: 784 = 784  ✓ valid
```


````python
import numpy as npA = np.array([[1, 2], [3, 4]])B = np.array([[5, 6], [7, 8]])print(f"A shape: {A.shape}")print(f"B shape: {B.shape}")print(f"A @ B shape: {(A @ B).shape}")# Visualize element-wise vs matrix multiplyprint("\nA * B (element-wise):")print(A * B)print("\nA @ B (matrix multiply):")print(A @ B)
````

````python
import matplotlib.pyplot as pltimport numpy as np# Heatmaps to show element-wise vs matrix multiplyfig, axes = plt.subplots(1, 2, figsize=(9, 4))axes[0].imshow(A * B, cmap='viridis')axes[0].set_title('Element-wise A * B', fontweight='bold')axes[1].imshow(A @ B, cmap='viridis')axes[1].set_title('Matrix multiply A @ B', fontweight='bold')for ax in axes:    ax.set_xticks([]); ax.set_yticks([])plt.tight_layout(); plt.show()
````

# Broadcasting

---

When you add a bias vector to a matrix of outputs, the shapes do not match. Broadcasting stretches the smaller array to fit.

```
| 1  2  3 |   +   [10, 20, 30]
| 4  5  6 |

| 11  22  33 |
| 14  25  36 |
```

# From Scratch to a Neural Network Layer

---

A single dense layer does `output = activation(W @ x + b)`:
- `W @ x` transforms input through the weight matrix
- `+ b` adds the bias (broadcasting)
- `activation()` applies non-linearity (ReLU, sigmoid...)

# Quiz Review

Try answering these before reading the explanations:

**Q1:** What's the rule for matrix multiplication shapes?
> **Explanation:** `(m × n) @ (n × p) = (m × p)` — inner dimensions must match.

**Q2:** What's the difference between `*` and `@` in NumPy?
> **Explanation:** `*` is element-wise; `@` is matrix multiplication (dot products of rows × columns).

**Q3:** What is broadcasting?
> **Explanation:** Stretching a smaller array to match a larger one by repeating along missing dimensions — how bias addition works.

# Key Terms

---

| Term | What people say | What it actually means |
|------|----------------|----------------------|
| Matrix multiply | Dot products between every row of the first matrix and every column of the second | Order matters |
| Transpose | Swap rows and columns | Critical in backpropagation |
| Broadcasting | Magic shape fixing | Stretching a smaller array to match a larger one |
| Identity matrix | The boring matrix | Multiplying by 1 — used in residual connections (ResNets) |
