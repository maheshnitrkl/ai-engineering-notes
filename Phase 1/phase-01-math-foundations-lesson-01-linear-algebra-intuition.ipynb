# Vectors Are Points (and Directions)

---

A vector is just a list of numbers, but those numbers are coordinates in space.

In AI, vectors represent everything:
- A **word** → a vector of 768 numbers (its "meaning" in embedding space)
- An **image** → a vector of millions of pixel values
- A **user** → a vector of preferences

The vector `[3, 2]` points from origin `(0,0)` to `(3, 2)` on the plane. Its magnitude is `sqrt(3² + 2²) = sqrt(13)`, pointing up and to the right.


````python
import numpy as npimport matplotlib.pyplot as plt# Plot vectors as arrows from originvectors = np.array([[3, 2], [1, 4], [-2, 1]])colors = ['#e74c3c', '#27ae60', '#3498db']labels = ['v1 = [3, 2]', 'v2 = [1, 4]', 'v3 = [-2, 1]']fig, ax = plt.subplots(figsize=(7, 6))for v, c, lab in zip(vectors, colors, labels):    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=c, width=0.005)    ax.text(v[0] + 0.15, v[1] + 0.15, lab, color=c, fontsize=10, fontweight='bold')ax.set_xlim(-3, 4); ax.set_ylim(-1, 5)ax.axhline(0, color='gray', lw=0.8); ax.axvline(0, color='gray', lw=0.8)ax.set_xlabel('x'); ax.set_ylabel('y')ax.set_title('Vectors as Arrows from the Origin', fontsize=12, fontweight='bold')ax.grid(True, alpha=0.3)ax.set_aspect('equal')plt.tight_layout(); plt.show()
````

# Matrices Are Transformations

---

A matrix transforms one vector into another — it can rotate, scale, stretch, or project.

In AI, matrices **are** the model:
- Neural network **weights** → matrices that transform input into output
- **Attention scores** → matrices that decide what to focus on
- **Embeddings** → matrices that map words to vectors

When you multiply a matrix by a vector, you're applying that transformation to the point.


# The Dot Product Measures Similarity

---

```
a · b = a₁×b₁ + a₂×b₂ + ... + aₙ×bₙ
```

Same direction: `a · b > 0` (similar)  
Perpendicular: `a · b = 0` (unrelated)  
Opposite direction: `a · b < 0` (dissimilar)

This is how search engines, recommendation systems, and RAG work — find vectors with high dot products.


````python
import numpy as np# Dot product = similaritya = np.array([1, 2, 3], dtype=float)b = np.array([4, 5, 6], dtype=float)c = np.array([1, 0, 0], dtype=float)print(f"a · b = {np.dot(a, b):.2f}"  )  # similar directionprint(f"a · c = {np.dot(a, c):.2f}"  )  # less aligned# Cosine similarity (normalized dot product)cos_ab = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))cos_ac = np.dot(a, c) / (np.linalg.norm(a) * np.linalg.norm(c))print(f"cosine(a, b) = {cos_ab:.4f}")print(f"cosine(a, c) = {cos_ac:.4f}")
````

# Linear Independence, Rank, and Projection

---

Vectors are **linearly independent** if no vector is a combination of the others.

**Rank** = number of linearly independent columns. Full rank → unique solution. Rank deficient → redundant features.

**Projection** of `a` onto `b`:
```
proj_b(a) = (a · b / b · b) × b
```

Projection is everywhere: linear regression, PCA, and attention all compute projections.

# Key Terms

---

| Term | What it means |
|------|---------------|
| Vector | A list of numbers representing a point or direction in n-dimensional space. |
| Matrix | A transformation that maps vectors from one space to another. |
| Dot product | A measure of how aligned two vectors are — the core of similarity search. |
| Linear independence | No vector in the set can be written as a combination of the others. |
| Rank | The number of linearly independent columns (or rows) in a matrix. |
| Projection | The component of one vector in the direction of another. |
