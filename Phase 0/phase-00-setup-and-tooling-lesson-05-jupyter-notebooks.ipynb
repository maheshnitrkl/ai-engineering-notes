# 02 Cell Types And Modes

---


A notebook is a list of **cells**. Each cell is either **code** or **markdown**.

## Two modes

| Mode | Visual cue | What you do |
|------|------------|-------------|
| **Command mode** | Blue bar on the left edge | Navigate, insert/delete cells, change cell type |
| **Edit mode** | Green bar on the left edge | Type code or text inside a cell |

- Press `Enter` → edit mode

# 03 Magic Commands

---


Magic commands are **Jupyter-specific**, not Python. They start with:

- `%` — **line magic** (affects one line)
- `%%` — **cell magic** (affects the whole cell)

---

## `%timeit` — Microbenchmarking

Runs the code many times and reports the average. Best for small, fast operations.


```python
import numpy as np

%timeit np.random.randn(10000)
```

**Typical output:**


```python
45.2 µs ± 1.3 µs per loop (mean ± std. dev. of 7 runs, 10000 loops each)
```

**When to use:** Comparing two ways of doing the same small task (e.g., list comprehension vs. NumPy).

---

## `%%time` — Wall clock timing

Runs the cell once and reports total elapsed time. Best for longer operations like model training.


```python
%%time
# Simulate a training loop
import time
time.sleep(2)
print("Done")
```

**Typical output:**


```python
Done
Wall time: 2.01 s
```

**Difference from `%timeit`:**
- `%timeit` → many runs, statistical average (microbenchmarks)
- `%%time` → single run, wall time (training runs, scripts)

---

## `%matplotlib inline` — Inline plots

Makes plots render directly inside the notebook, below the code cell.


```python
%matplotlib inline

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.plot([1, 2, 3, 4], [1, 4, 2, 3])
plt.title("Inline Plot")
plt.show()
```

The plot appears immediately under the cell. Without this, plots may open in a separate window or not appear at all.

---

## `!pip install` — Install packages without leaving the notebook

The `!` prefix runs any shell command.


```python
!pip install scikit-learn
```

You can also `!ls`, `!pwd`, `!curl`, etc.

> **Tip:** After installing, you may need to restart the kernel for the new package to be importable.

---

## `%env` — Check environment variables


```python
%env CUDA_VISIBLE_DEVICES
```

To set an environment variable inside the notebook:


```python
%env CUDA_VISIBLE_DEVICES=0
```

# 05 Kernel And State

---


## What is the kernel?

The **kernel** is a separate Python process running in the background. When you run a cell:

1. The notebook UI sends the code to the kernel
2. The kernel executes it
3. The kernel sends the result back to the UI
4. All output appears below the cell


```python
Notebook UI  <-->  Kernel (Python process)
                        |
                        v
              Keeps variables in memory
                        |
                        v
              Runs cells in whatever order you click
                        |
                        v
              Dies when you restart it
```

## Variable persistence

All cells share the same kernel, so **variables persist between cells**.

**Cell 1:**


```python
learning_rate = 0.01
epochs = 10
```

**Cell 2 (runs later):**


```python
print(f"Training for {epochs} epochs at lr={learning_rate}")
# Output: Training for 10 epochs at lr=0.01
```

## Restarting the kernel

| Action | What happens |
|--------|--------------|
| **Restart** | Clears all variables; code stays in cells |
| **Restart & Run All** | Clears variables, then runs every cell top-to-bottom |
| **Re-run all cells** | Keeps variables, runs every cell again |

**When to restart:**
- After installing a new package with `!pip install`
- When you suspect hidden state is causing bugs
- Before sharing a notebook with someone else
- When memory usage is high

## State inspection

See what's currently in memory:


```python
# List all variables in the current namespace
%who

# List with types
%who_ls

# Detailed info
%whos
```

# 06 Notebooks Vs Scripts

---


## When to use which

| Use notebooks for | Use scripts (`.py`) for |
|-------------------|-------------------------|
| Exploring a dataset | Training pipelines |
| Prototyping a model | Reusable utilities |
| Visualizing results | Anything with `if __name__ == "__main__":` |
| Explaining your work | Code that runs on a schedule |
| Quick experiments | Production code |
| Course exercises | Packages and libraries |

## The golden rule

> **Explore in notebooks, ship in scripts.**

## Common AI workflow

1. **Explore data** in a notebook — load CSVs, check distributions, plot charts
2. **Prototype the model** in the notebook — try architectures, tune hyperparameters
3. **Move working code** to `.py` files — data loaders, model definitions, training loops
4. **Import those `.py` files** back into the notebook for further experiments


```python
Notebook (exploration)
    ↓  (code that works)
.py script (production)
    ↓  (import back)
Notebook (further experiments using the script)
```

## Why not use notebooks for everything?

- **Version control is harder** — `.ipynb` files are JSON and diff poorly in Git
- **Hidden state** — variables from out-of-order cells can break reproducibility
- **No `if __name__ == "__main__":`** — notebooks don't have a clean entry point
- **Harder to test** — you can't easily `import` a notebook and unit-test its functions

# 08 Common Traps And Fixes

---


Notebooks are powerful but have real pitfalls. Here are the three most common ones.

## Trap 1: Out-of-order execution

**The problem:** You run cell 5, then cell 2, then cell 7. The notebook works on your machine because variables exist in memory, but it breaks when someone runs it top-to-bottom.

**Example of the bug:**

Cell 1:


```python
x = 5
```

Cell 2:


```python
y = x + 10  # Works if you ran Cell 1 first
```

If you run Cell 2 before Cell 1, you get `NameError: name 'x' is not defined`.

**The fix:** Before sharing, use **Kernel > Restart & Run All**. This clears memory and runs every cell in order, proving the notebook is self-contained.

---

## Trap 2: Hidden state

**The problem:** You delete a cell, but the variable it created still exists in the kernel's memory. The notebook looks clean, but it depends on a "ghost cell."

**Example:**
1. Run a cell that defines `model = train_model()`
2. Delete that cell
3. Run a later cell that uses `model` — it still works because `model` is in memory

**The fix:** **Restart the kernel regularly**, especially after deleting cells. Use **Kernel > Restart & Run All** to verify everything works from scratch.

---

## Trap 3: Memory leaks

**The problem:** You load a 4 GB dataset, train a model, then load another dataset. Nothing gets freed automatically. Memory usage keeps climbing until the kernel crashes.

**Example:**


```python
# Cell 1
data = pd.read_csv("huge_dataset.csv")  # 4 GB

# Cell 2
model = train(data)

# Cell 3
data2 = pd.read_csv("another_dataset.csv")  # Another 4 GB — `data` still in memory!
```

**The fix:**


```python
import gc

del data          # Remove the reference
gc.collect()      # Force garbage collection
```

Or simply **restart the kernel** between major tasks.

---

- [ ] Kernel > Restart & Run All (no errors)
- [ ] All outputs visible and correct
- # No `!pip install` cells that haven't been run
- # No variables depending on deleted cells
- # File paths are relative (not `C:\Users\...\data.csv`)

# 09 Key Terms Glossary

---


| Term | What people say | What it actually means |
|------|----------------|------------------------|
| **Kernel** | "The thing running my code" | A separate Python process that executes cells and keeps variables in memory. All cells in a notebook share the same kernel. |
| **Cell** | "A code block" | An independently runnable unit in a notebook. Either code (Python) or markdown (formatted text). |
| **Magic command** | "Jupyter tricks" | Special commands prefixed with `%` (line magic) or `%%` (cell magic) that control the notebook environment. Examples: `%timeit`, `%%time`, `%matplotlib inline`. |
| **`.ipynb`** | "Notebook file" | A JSON file containing cells, outputs, and metadata. Stands for **IPython Notebook**. Can be opened in JupyterLab, Jupyter Notebook, or VS Code. |
| **Command mode** | "Navigation mode" | Cell-level navigation (blue sidebar). Press `Esc` to enter. Use `A`, `B`, `D D`, `M`, `Y`, `Shift+Enter`. |
| **Edit mode** | "Typing mode" | Cell content editing (green sidebar). Press `Enter` to enter. Use `Tab`, `Shift+Tab`, `Ctrl+/`. |
| **Inline output** | "Output below the cell" | Results (text, plots, tables) displayed directly underneath the code cell that produced them. |
| **Restart & Run All** | "Reset everything" | Clears the kernel's memory and runs every cell top-to-bottom. Used to verify reproducibility. |

## Exercise 2: Load CSV, display dataframe, plot chart

**Goal:** Create a notebook with both markdown and code cells that loads data, shows it, and visualizes it. Then verify with Restart & Run All.


```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create sample data (no external CSV needed)
np.random.seed(42)
df = pd.DataFrame({
    "epoch": range(1, 21),
    "loss": np.exp(-np.linspace(0, 3, 20)) + np.random.normal(0, 0.02, 20),
    "accuracy": 1 - np.exp(-np.linspace(0, 3, 20)) + np.random.normal(0, 0.02, 20)
})

# Display the dataframe (rich HTML output)
df.head()

# Plot loss curve
plt.figure(figsize=(10, 4))
plt.plot(df["epoch"], df["loss"], marker="o", label="Loss")
plt.plot(df["epoch"], 1 - df["accuracy"], marker="s", label="Error rate")
plt.xlabel("Epoch")
plt.ylabel("Value")
plt.title("Training Metrics")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Question 1 (Pre)
**What is a Jupyter notebook?**

---

## Question 5 (Post)
**When should you move code from a notebook into a `.py` script?**

> **Correct answer:** When the code is a reusable utility, training pipeline, or production code.

The rule is **"explore in notebooks, ship in scripts."** Notebooks are for exploration and prototyping; `.py` files are for reusable, schedulable, production code.

## Common traps & fixes

| Trap | Fix |
|------|-----|
| Out-of-order execution | Kernel > Restart & Run All |
| Hidden state (ghost variables) | Restart kernel regularly |
| Memory leaks | `del var; gc.collect()` or restart kernel |

## Notebooks vs Scripts

- **Notebooks:** Exploration, prototyping, visualization, teaching
- **Scripts (`.py`):** Production, pipelines, reusable utilities, scheduled tasks
- **Rule:** Explore in notebooks, ship in scripts

## File format

- `.ipynb` = JSON file with cells, outputs, metadata
- All interfaces (JupyterLab, Jupyter Notebook, VS Code) use the same format
