# Why Version Control Matters

---


You're about to write hundreds of code files across 20 phases. Without version control:

- You **lose work** when you accidentally delete or break something
- You **can't undo** changes that introduced a bug
- You **can't collaborate** with others on the same codebase

Git tracks every change to every file. It's like an infinite undo button for your entire project.
GitHub is where your git repository lives in the cloud — a backup you can access from any computer.


# Visual: The 'Final_v2_REALLY_FINAL' Problem

---


````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
ax.axis('off')

files = [
    'model.py',
    'model_v2.py',
    'model_v2_fixed.py',
    'model_v2_fixed_REALLY_FINAL.py',
    'model_v2_fixed_REALLY_FINAL_backup.py',
]

table_data = [[f] for f in files]
table = ax.table(
    cellText=table_data,
    colLabels=['File name'], loc='center', cellLoc='left',
    colColours=['#e74c3c']
)
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1, 2)
for i in range(1):
    table[(0, i)].set_text_props(color='white', fontweight='bold')

plt.title("Without Version Control: File Naming Chaos", fontsize=12, fontweight="bold", pad=20)
plt.tight_layout()
plt.show()
````

> **With git,** you have ONE file (`model.py`) and git remembers every version of it. You never need `v2_final_REALLY_FINAL.py` again.


# The Concept: Four Areas Git Moves Changes Through

---


Your changes flow through four areas. Think of it like preparing a document:

| Area | Analogy | Git Command |
|------|---------|-------------|
| Working Directory | Your desk, where you edit | `git status` |
| Staging Area | The photocopier, where you select what to save | `git add` |
| Local Repo | The filing cabinet, where saved versions live | `git commit` |
| Remote (GitHub) | The cloud backup, accessible from anywhere | `git push` / `git pull` |

```mermaid
sequenceDiagram
    participant WD as Working Directory
    participant SA as Staging Area
    participant LR as Local Repo
    participant R as Remote (GitHub)
    WD->>SA: git add
    SA->>LR: git commit
    LR->>R: git push
    R->>LR: git fetch
    LR->>WD: git pull
```


# Step 1: Configure Git Identity

---


Git needs to know who made each change. This is attached to every commit.

```bash
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
```

Verify:
```bash
git config --global --list
```

Without this, git refuses to commit because it doesn't know how to label your work.


# Step 2: The Daily Workflow — add, commit, push

---


This is the loop you'll repeat dozens of times per day:

```bash
git status          # See what changed
git add file.py     # Stage the changes you want to save
git commit -m "Add perceptron implementation"  # Save a snapshot with a message
git push origin main  # Upload to GitHub backup
```

| Step | Command | What it actually does |
|------|---------|----------------------|
| 1 | `git status` | Shows which files are modified, new, or deleted |
| 2 | `git add <file>` | Copies changes into the staging area (prepares them for saving) |
| 3 | `git commit -m "msg"` | Creates a permanent snapshot in your local repo |
| 4 | `git push origin main` | Uploads your snapshots to GitHub |


# Visual: The Commit Sequence

---


````python
import matplotlib.pyplot as plt

steps = ['git status\n(see changes)', 'git add\n(stage)', 'git commit\n(save)', 'git push\n(backup)']
colors = ['#f39c12', '#e67e22', '#27ae60', '#2c3e50']

fig, ax = plt.subplots(figsize=(9, 3))
for i, (step, color) in enumerate(zip(steps, colors)):
    ax.annotate('', xy=(i+1, 0), xytext=(i, 0),
                arrowprops=dict(arrowstyle='->', color=color, lw=3))
    ax.text(i+0.5, 0.3, step, ha='center', va='bottom', fontsize=10,
            fontweight='bold', color=color)

ax.set_xlim(-0.3, 3.7)
ax.set_ylim(0, 1)
ax.axis('off')
ax.set_title("The Daily Git Workflow", fontsize=12, fontweight="bold", pad=20)
plt.tight_layout()
plt.show()
````

> **Order matters:** You can't commit before you add, and you can't push before you commit.


# Step 3: Branching — Parallel Universes for Experiments

---


The `main` branch should always be stable. When you want to try something risky, create a branch.

Think of a branch as a **parallel universe** — you can change anything there without affecting `main`.

```bash
# Create and switch to a new branch
git checkout -b experiment/new-optimizer

# Make changes, test, commit
git add .
git commit -m "Try new optimizer"

# Switch back to main
git checkout main

# Merge the experiment into main
git merge experiment/new-optimizer
```

If the experiment fails, abandon it:
```bash
git checkout main
git branch -D experiment/new-optimizer
```
`main` is untouched — the failed experiment is gone.


# Visual: Branch, Experiment, Merge

---


````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))
ax.axis('off')

# Main branch line
main_x = [0, 1, 2, 3, 4]
main_y = [2] * 5
ax.plot(main_x, main_y, 'o-', color='#2c3e50', linewidth=4, markersize=11, label='main')

# Feature branch (diverges at main commit 1)
feat_x = [1, 1.5, 2.5, 3.5]
feat_y = [2, 4, 4, 4]
ax.plot(feat_x, feat_y, 'o-', color='#e74c3c', linewidth=4, markersize=11, label='feature branch')

# Merge back
ax.plot([3.5, 4], [4, 2], 'o-', color='#27ae60', linewidth=4, markersize=11)

# Annotations
ax.annotate('Branch created\nfrom main', xy=(1, 2.8), ha='center', fontsize=9, color='#e74c3c')
ax.annotate('Experiment\ncommits', xy=(2.5, 4.4), ha='center', fontsize=9, color='#e74c3c')
ax.annotate('Merged back\nto main', xy=(4, 1.4), ha='center', fontsize=9, color='#27ae60')
ax.annotate('main\n(stable)', xy=(0, 1.4), ha='center', fontsize=9, color='#2c3e50')

ax.set_xlim(-0.5, 4.5)
ax.set_ylim(0.5, 5)
ax.set_title("Branching: Experiment Without Breaking main", fontsize=12, fontweight="bold", pad=20)
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
````

> **Rule:** `main` stays stable. All experiments happen on branches.

````python
import matplotlib.pyplot as plt

weeks = ['Week 1', 'Week 2', 'Week 3', 'Week 4', 'Week 5', 'Week 6']
commits = [3, 8, 15, 22, 31, 40]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(weeks, commits, 'o-', color='#2c3e50', linewidth=3, markersize=8)
ax.fill_between(weeks, commits, alpha=0.3, color='#2c3e50')

ax.set_ylabel("Total Commits", fontsize=11)
ax.set_title("Commits Accumulate as You Progress Through Lessons", fontsize=12, fontweight="bold")
ax.grid(True, alpha=0.3, axis='y')

for i, c in enumerate(commits):
    ax.text(i, c + 0.8, str(c), ha="center", fontweight="bold")

plt.tight_layout()
plt.show()
````

# Why .gitignore Matters for AI Projects

---


Git is designed for text files (code). It's terrible at large binary files.

Model checkpoints like `.pt`, `.pth`, `.safetensors` can be **gigabytes** each. If you commit them, your repo becomes enormous and slow.

Add them to `.gitignore` so git ignores them entirely.

```
# Model checkpoints
*.pt
*.pth
*.safetensors
*.bin

# Virtual environments
.venv/
venv/

# OS files
.DS_Store
Thumbs.db

# Python
*.pyc
__pycache__/
```


# Visual: Why Model Files Don't Belong in Git

---


````python
import matplotlib.pyplot as plt

file_types = ['Source Code (.py)', 'Config (.yaml)', 'Model (.pt)', 'Model (.pth)', 'Weights (.safetensors)']
sizes_mb = [0.005, 0.002, 500, 2000, 5000]
colors = ['#27ae60', '#27ae60', '#e74c3c', '#e74c3c', '#e74c3c']

fig, ax = plt.subplots(figsize=(9, 4))
bars = ax.bar(file_types, sizes_mb, color=colors, alpha=0.8)
ax.set_ylabel("Size (MB, log scale)", fontsize=11)
ax.set_title("Typical File Sizes: Code vs Model Checkpoints", fontsize=12, fontweight="bold")
ax.set_yscale('log')
ax.grid(True, alpha=0.3, axis='y')
plt.xticks(rotation=15, ha='right')

for bar, mb in zip(bars, sizes_mb):
    ax.text(bar.get_x() + bar.get_width()/2, mb * 1.1, f"{mb:,.0f} MB", ha="center",
            fontweight="bold", fontsize=9)

plt.tight_layout()
plt.show()
````

> **Source code:** kilobytes. **Model checkpoints:** gigabytes. Git handles the first well, chokes on the second.

**Q1:** What does 'version control' primarily help you do?
> **Explanation:** Track changes to files over time and collaborate with others. Git records every change so you can revert to any previous state.

**Q2:** What is a 'repository'?

> **Explanation:** Creates a new branch and switches to it immediately. You can experiment freely without touching `main`.

**Q5:** Why add `.pt`, `.pth`, `.safetensors` to `.gitignore`?
> **Explanation:** These are model checkpoint files that are gigabytes in size. Git is not designed for large binary files — they should be regenerated or stored separately.

# Key Terms

---


| Term | What people say | What it actually means |
|------|----------------|----------------------|
| Commit | "Saving" | A snapshot of your entire project at a point in time |
| Branch | "A copy" | A pointer to a commit; moves forward as you work |
| Merge | "Combining code" | Taking changes from one branch and applying them to another |
| Remote | "The cloud" | A copy of your repo hosted on GitHub/GitLab |
| Staging Area | "Preparing" | The intermediate area where you select changes before committing |
